# Extensión de la ocurrencia (EOO)

> **Extensión de la ocurrencia (EOO).** La EOO de un ecosistema es el área (km2) de un polígono convexo mínimo – el polígono más pequeño en el que ningún ángulo interno supera los 180° y que abarca todas las ocurrencias espaciales actuales conocidas del tipo de ecosistema. 
> 
> El polígono convexo mínimo (también conocido como envolvente convexa) no debe excluir ninguna área, discontinuidad o disyunción, independientemente de si el ecosistema puede ocurrir o no en esas áreas. Regiones como océanos (para ecosistemas terrestres), tierra firme (para ecosistemas costeros o marinos), o áreas fuera del área de estudio (como en un país diferente) deben permanecer incluidas dentro del polígono convexo mínimo para garantizar que este método estandarizado sea comparable entre tipos de ecosistemas. Además, estas características contribuyen a distribuir los riesgos a lo largo de la distribución del ecosistema, al hacer que las diferentes partes de su distribución sean más independientes espacialmente.

::: {.notes}
<!-- en-source: auto-generated speaker notes, do not edit -->
**Extent of occurrence (EOO)**

> **Extent of occurrence (EOO).** The EOO of an ecosystem is the area (km2) of a minimum convex polygon – the smallest polygon in which no internal angle exceeds 180° that encompasses all known current spatial occurrences of the ecosystem type.
>
> The minimum convex polygon (also known as a convex hull) must not exclude any areas, discontinuities or disjunctions, regardless of whether the ecosystem can occur in those areas or not. Regions such as oceans (for terrestrial ecosystems), land (for coastal or marine ecosystems), or areas outside the study area (such as in a different country) must remain included within the minimum convex polygon to ensure that this standardised method is comparable across ecosystem types. In addition, these features contribute to spreading risks across the distribution of the ecosystem by making different parts of its distribution more spatially independent.
:::

# Área de ocupación (AOO)

::: {.notes}
<!-- en-source: auto-generated speaker notes, do not edit -->
**Area of occupancy (AOO)**
:::


::: {.scrollbox}

> **Área de ocupación (AOO). La AOO de un ecosistema definido en la RLE se determina contando el número de celdas de una cuadrícula de 10 × 10 km^2^ que contienen el tipo de ecosistema.**

> Las estimaciones de la extensión del ecosistema son altamente sensibles al tamaño de grano (resolución espacial) de los mapas fuente (Nicholson et al., 2009), por lo que todas las medidas de AOO de un tipo de ecosistema deben estandarizarse a un tamaño de grano espacial común (Keith et al., 2018). Este tamaño de grano relativamente grande se aplica por cuatro razones:
>
> (i) los límites de los ecosistemas son inherentemente vagos (Regan et al., 2002), por lo que es más fácil determinar que una ocurrencia de un ecosistema se encuentra dentro de una celda de cuadrícula más grande que en una más pequeña;
> (ii) pueden requerirse celdas más grandes para diagnosticar la presencia de ecosistemas caracterizados por procesos que operan a grandes escalas espaciales, o que poseen rasgos diagnósticos escasos, crípticos, agrupados o móviles (por ejemplo, sistemas pelágicos o artesianos);
> (iii) las celdas más grandes permiten estimar la AOO incluso cuando los datos de distribución de alta resolución son limitados; y
> (iv) los estudios de simulación han indicado que las celdas más grandes predicen mejor el riesgo frente a eventos de amenaza del mundo real que las celdas de escala más fina (Keith et al., 2018). Un conjunto de datos global en cuadrícula de 10 × 10 km adecuado para este propósito está disponible a través de un repositorio de datos público en formatos raster y vectorial (Murray, 2017).

> Algunas distribuciones de ecosistemas presentan una distribución de tamaños de parche muy sesgada. En estos casos, un gran número de parches pequeños contribuye con un efecto de dispersión de riesgo insignificante en comparación con el de los parches más grandes. Por lo tanto, los evaluadores deben aplicar una corrección excluyendo de la AOO aquellas celdas de la cuadrícula que, en conjunto, contienen parches del tipo de ecosistema que representan menos del 1% del área total mapeada del tipo de ecosistema, incluyendo así siempre el 99% de la extensión del ecosistema.
>
> El protocolo para este ajuste incluye los siguientes pasos:
>
>   (1) Intersectar la cuadrícula de AOO con el mapa de distribución del ecosistema.
>   (2) Calcular la extensión del tipo de ecosistema en cada celda de la cuadrícula ('área') y sumar estas áreas para obtener el área total del ecosistema ('área total').
>   (3) Ordenar las celdas de la cuadrícula en orden ascendente según su área (las más pequeñas primero). Calcular la suma acumulada del área por celda ('área acumulada').
>   (4) Calcular la 'proporción acumulada' dividiendo el 'área acumulada' entre el 'área total' (la proporción acumulada toma valores entre 0 y 1).
>   (5) Calcular la AOO contando el número de celdas con una 'proporción acumulada' mayor a 0.01 (es decir, excluir las celdas que en conjunto representan hasta el 1% de la extensión total mapeada del tipo de ecosistema).

:::

::: {.notes}
<!-- en-source: auto-generated speaker notes, do not edit -->
> **Area of occupancy (AOO). The AOO of an ecosystem defined in the RLE is determined by counting the number of 10 × 10 km^2^ grid cells that contain the ecosystem type.**

> Estimates of ecosystem extent are highly sensitive to the grain size (spatial resolution) of source maps (Nicholson et al., 2009), so all measures of AOO of an ecosystem type must be standardised to a common spatial grain (Keith et al., 2018). This relatively large grain size is applied for four reasons:
>
> (i) ecosystem boundaries are inherently vague (Regan et al., 2002), so it is easier to determine that an ecosystem occurrence falls within a larger grid cell than a smaller one;
> (ii) larger cells may be required to diagnose the presence of ecosystems characterised by processes that operate over large spatial scales, or possess diagnostic features that are sparse, cryptic, clustered or mobile (e.g. pelagic or artesian systems);
> (iii) larger cells allow AOO estimation even when high resolution distribution data are limited; and
> (iv) simulation studies have indicated that larger cells better predict risk in the face of real-world threat events than finer scale cells (Keith et al., 2018). A global 10 × 10 km gridded dataset suitable for this purpose is available via a public data repository in raster and vector formats (Murray, 2017).
>
> Some ecosystem distributions comprise a highly skewed distribution of patch sizes. In these cases, large numbers of small patches contribute a negligible risk-spreading effect compared to that of larger patches. Assessors should therefore apply a correction by excluding from the AOO those grid cells that, collectively, contain patches of the ecosystem type that account for less than 1% of the total mapped area of the ecosystem type, thus always including 99% of the ecosystem extent.
>
> The protocol for this adjustment includes the following steps:
>
> (1) Intersect AOO grid with the ecosystem’s distribution map.
> (2) Calculate extent of the ecosystem type in each grid cell (‘area’) and sum these areas to obtain the total ecosystem area (‘total area’).
> (3) Arrange grid cells in ascending order based on their area (smaller first). Calculate accumulated sum of area per cell (‘cumulative area’).
> (4) Calculate ‘cumulative proportion’ by dividing ‘cumulative area’ by ‘total area’ (cumulative proportion takes values between 0 and 1)
> (5) Calculate AOO by counting the number of cells with a ‘cumulative proportion’ greater than 0.01 (i.e. exclude cells that in combination account for up to 1% of the total mapped extent of the ecosystem type).
:::

```
## AOO Calculation Steps

::: {.scrollbox}
```

::: {.notes}
<!-- en-source: auto-generated speaker notes, do not edit -->
**AOO Calculation Steps**
:::


````python
#| echo: true
keep = ['geometry', 'grid_col', 'grid_row', ecosystem_column]
gdf = aoo_grid_filtered.grid_cells[keep].sort_values(by=ecosystem_column)   # <3>
gdf["cumulative_area"] = gdf[ecosystem_column].cumsum()
total_area = gdf["cumulative_area"].iloc[-1]                                # <4>
gdf["cumulative_proportion"] = gdf["cumulative_area"] / total_area          # <4>
display(gdf)

aoo = len(gdf[gdf["cumulative_proportion"] > 0.01])   # <5>
print(f'AOO is {aoo} cells')
````
